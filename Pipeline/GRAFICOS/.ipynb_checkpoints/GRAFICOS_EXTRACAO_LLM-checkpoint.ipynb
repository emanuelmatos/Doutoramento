```python
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import re
from tqdm import tqdm # Para barras de progresso, útil em operações longas

# ========================
# Diretórios e configurações
# ========================
working_dir = os.getcwd()
print(working_dir)

parts = working_dir.split("/")

#if parts[-1] == "pipeline":
if parts[-1].startswith("pipeline"): ### BUG SOLVED MAY 2025
    pipeline = working_dir
else:
    pipeline ="/".join(parts[:-1])
print(pipeline)

base = pipeline
sys.path.append(base)

from config.configsite import *
# ## Configuração de Caminhos e Carregamento de Dados

##################################################################

# Define o diretório base para os arquivos de entrada/saída
# Utilize os.getcwd() e manipulação para tornar o caminho mais robusto
working_dir = os.getcwd()

# Ajusta o caminho base dependendo da estrutura do diretório
if os.path.basename(working_dir).startswith("pipeline"):
    # Se o diretório atual já é o pipeline ou um subdiretório direto
    pipeline_base_dir = working_dir
else:
    # Caso contrário, assume que o script está um nível abaixo do 'pipeline'
    pipeline_base_dir = os.path.join(working_dir, '..') # Ajuste conforme a sua estrutura real se necessário

# Definindo o diretório de resultados, assumindo uma estrutura comum
# Ex: /home/ematos/phd/pipeline_city/ENTITY_DETECTOR_BY_BERT/OUTPUT_RESULTS/
output_results_dir = os.path.join(pipeline_base_dir, 'NEW_INDOMAIN_NER_LLM', 'OUTPUT_RESULTS')

# Nome do arquivo de resultados LLM
arquivo = f"XML_2PerClass_10class_sabiazinho3_{dataset}_pt_512_20250519.txt"
filename_results = f"XML_2PerClass_10class_sabiazinho3_{dataset}_pt_512_20250519.txt"
file_path_results = os.path.join(output_results_dir, filename_results)


# Carregar o dataset
try:
    # O seu snippet original usava 'dat = pd.read_csv(base+fileresults, sep='\t')'
    # E depois 'df = dat' dentro da função. É mais limpo passar o DataFrame diretamente.
    # Ou, se o arquivo for sempre o mesmo, carregá-lo globalmente ou passá-lo para a função.
    # Vamos carregar o DataFrame aqui e passá-lo para a função para maior flexibilidade.
    df = pd.read_csv(file_path_results, sep='\t')
    print("Dataset carregado com sucesso!")
    print(f"Formato do dataset: {df.shape}")
    print("Primeiras 5 linhas do dataset:")
    print(df.head())
except FileNotFoundError:
    print(f"Erro: O arquivo '{file_path_results}' não foi encontrado. Verifique o caminho.")
    df = None # Define como None para evitar erros posteriores


# ======================
# 2. Extrair tags e termos de cada linha
# ======================
def extrair_tags_termos(texto):
    """
    Retorna uma lista de (tag, termo) extraídos do texto marcado com XML.
    """
    padrao = re.findall(r'<([^<>]+)>([^<>]+)</\1>', texto)
    return padrao if padrao else [(None, None)]

# Aplica a extração a cada linha
df['extraido'] = df['frase_tag'].apply(extrair_tags_termos)

# ======================
# 3. Explodir e separar
# ======================
df_explodido = df.explode('extraido', ignore_index=True)
df_explodido[['tag', 'termo']] = pd.DataFrame(df_explodido['extraido'].tolist(), index=df_explodido.index)

# ======================
# 4. Resultado final com frase_tag preservada
# ======================
df_final = df_explodido[['linha', 'frase', 'frase_tag', 'tag', 'termo']]
df_final = df_final[df_final['tag'].notna()]  # Remove onde não há tag

# ======================
# 5. Visualizar
# ======================
print(df_final.head(10))

# ======================
# 6. (Opcional) Salvar em CSV
# ======================
#df_final.to_csv("tags_extraidas_completas.csv", index=False)
```

```python
import matplotlib.pyplot as plt

# ============================================
# 1. Contagem por tag (ocorrência de termos)
# ============================================
contagem_termos = df_final['tag'].value_counts().sort_values(ascending=False)
termos_total = contagem_termos.sum()
termos_pct = contagem_termos.cumsum() / termos_total

# Mantém apenas as tags que acumulam até 80%
tags_80_termos = termos_pct[termos_pct <= 0.80].index
contagem_termos_80 = contagem_termos[tags_80_termos]

# ============================================
# 2. Contagem por linhas distintas com a tag
# ============================================
contagem_linhas = df_final.groupby('tag')['linha'].nunique().sort_values(ascending=False)
linhas_total = contagem_linhas.sum()
linhas_pct = contagem_linhas.cumsum() / linhas_total

tags_80_linhas = linhas_pct[linhas_pct <= 0.80].index
contagem_linhas_80 = contagem_linhas[tags_80_linhas]

# ============================================
# 3. Gráfico: 80% dos termos
# ============================================
plt.figure(figsize=(12, 6))
contagem_termos_80.plot(kind='bar', color='#2E8B57')

plt.title('Tags que Representam 80% dos Termos Anotados')
plt.xlabel('Tag')
plt.ylabel('Número de Termos')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()

# ============================================
# 4. Salvar e exibir o gráfico
# ============================================
output_graph_filename = f'top_80_perc_bar_chart_LLM_{dataset}.png'
output_graph_path = os.path.join(output_graph_filename)

plt.savefig(output_graph_path, bbox_inches='tight')
print(f"\n📁 Gráfico salvo em: {output_graph_path}")


# ============================================
# 4. Gráfico: 80% das linhas distintas
# ============================================
plt.figure(figsize=(12, 6))
contagem_linhas_80.plot(kind='bar', color='#4682B4')
plt.title('Tags que Ocorreram em 80% das Linhas Distintas')
plt.xlabel('Tag')
plt.ylabel('Número de Linhas')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# ======================================
# 1. Contar combinações tag + termo
# ======================================
contagem_tag_termo = df_final.groupby(['tag', 'termo']).size().reset_index(name='frequencia')
contagem_tag_termo_sorted = contagem_tag_termo.sort_values(by='frequencia', ascending=False)

# ======================================
# 2. Criar rótulo combinado para visualização
# ======================================
contagem_tag_termo_sorted['tag_termo'] = contagem_tag_termo_sorted['tag'] + ' | ' + contagem_tag_termo_sorted['termo']

# ======================================
# 3. Selecionar top N combinações
# ======================================
top_n = 20
top_comb = contagem_tag_termo_sorted.head(top_n)

# ======================================
# 4. Gráfico de barras
# ======================================
plt.figure(figsize=(14, 6))
plt.bar(top_comb['tag_termo'], top_comb['frequencia'], color='#6A5ACD')
plt.title(f'Top {top_n} Combinações TAG vs TERMO mais Frequentes')
plt.xlabel('TAG | TERMO')
plt.ylabel('Frequência')
plt.xticks(rotation=75, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()



# ============================================
# 4. Salvar e exibir o gráfico
# ============================================
output_graph_filename = f'top_{top_n}_combinacoes_bar_chart_LLM_{dataset}.png'
output_graph_path = os.path.join(output_graph_filename)

plt.savefig(output_graph_path, bbox_inches='tight')
print(f"\n📁 Gráfico salvo em: {output_graph_path}")


# ======================================
# 5. Tabela completa com todas as combinações
# ======================================
tabela_completa = contagem_tag_termo_sorted[['tag', 'termo', 'frequencia']]

# Salva como Excel
output_excel = f"tabela_TAG_vs_TERMO_frequencia_{dataset}.xlsx"
tabela_completa.to_excel(output_excel, index=False)

print(f"📁 Tabela salva em: {output_excel}")
```
