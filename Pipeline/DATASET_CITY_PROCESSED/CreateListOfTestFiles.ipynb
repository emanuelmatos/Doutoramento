```python
def get_list_of_files(directory, pattern= "para"):  # pattern is only the start!!!
    print(f"List files {pattern} in {directory}")
    matches = []
    for root, dirs, files in os.walk(directory):
        for file in files:
            if file.startswith(pattern):
                matches.append(os.path.join(root, file))
    
    #print(matches)
    return matches
```

```python

def create_csv_for_files(directory, output_csv):
    # Get all filenames in the specified directory
    filenames = get_list_of_files(directory)  #os.listdir(directory)
    print(filenames[:3])
    
    # Create the CSV file
    with open(output_csv, 'w', encoding='utf-8', newline='') as csvfile:
        csv_writer = csv.writer(csvfile)
        
        # Write header row
        csv_writer.writerow(['filename', 'basename'])
        
        # Write each filename and a unique ID
        for idx, filename_complete in enumerate(filenames, start=1):

            parts = filename_complete.split("/")

            #print (parts)
            filename = parts[-1]  #+".txt"

            print(f"{filename}, {filename[:-4]}")
            csv_writer.writerow([filename, filename[:-4]])
    
    print(f"CSV file created: {output_csv} at {output_dir}")

# Example usage
create_csv_for_files(f'{pipeline}{dataset}/', filename)

```
