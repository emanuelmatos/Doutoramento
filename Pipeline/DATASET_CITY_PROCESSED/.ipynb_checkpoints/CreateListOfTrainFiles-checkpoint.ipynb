```python
import os
import random # Importa o módulo random para embaralhar a lista

def criar_fila_de_nomes_txt(caminho_da_pasta_origem, caminho_da_pasta_destino,
                            nome_arquivo_principal="files_list.csv",
                            nome_arquivo_teste="test_files_list.csv",
                            nome_arquivo_original="files_list_original.csv", # Novo parâmetro para o nome do arquivo original
                            percentual_teste=0.30):
    """
    Lê os nomes de todos os arquivos .txt em uma pasta de origem especificada,
    separa uma porcentagem para um arquivo de teste e o restante para um arquivo principal,
    salvando ambos dentro de uma pasta de destino específica.
    Além disso, cria um terceiro arquivo com a lista completa e original de todos os nomes.
    Cada linha dos arquivos de saída terá o nome completo do arquivo e,
    separado por vírgula, o nome do arquivo sem a extensão ".txt".

    Args:
        caminho_da_pasta_origem (str): O caminho para a pasta que contém os arquivos .txt (ex: DATASET_CITY).
        caminho_da_pasta_destino (str): O caminho para a pasta onde os arquivos de saída serão salvos (ex: DATASET_CITY_PROCESSED).
        nome_arquivo_principal (str): O nome do arquivo onde a maioria dos nomes dos arquivos será salva.
        nome_arquivo_teste (str): O nome do arquivo onde a porcentagem de teste dos nomes dos arquivos será salva.
        nome_arquivo_original (str): O nome do arquivo onde a lista completa e original será salva.
        percentual_teste (float): A porcentagem (entre 0 e 1) dos arquivos a serem destinados ao conjunto de teste.
    """
    todos_nomes_dos_arquivos = []
    files_list_original_data = [] # Renomeada para evitar confusão com o arquivo
    try:
        # 1. Listar todos os arquivos .txt na pasta de origem e popular a lista original de dados
        for nome_arquivo in os.listdir(caminho_da_pasta_origem):
            caminho_completo_arquivo = os.path.join(caminho_da_pasta_origem, nome_arquivo)
            if os.path.isfile(caminho_completo_arquivo) and nome_arquivo.endswith(".txt"):
                todos_nomes_dos_arquivos.append(nome_arquivo) # Esta lista será embaralhada
                nome_sem_extensao = os.path.splitext(nome_arquivo)[0]
                files_list_original_data.append((nome_arquivo, nome_sem_extensao)) # Adiciona à lista original de dados

        # 2. Criar a pasta de destino se ela não existir
        os.makedirs(caminho_da_pasta_destino, exist_ok=True)
        print(f"📁 Garantindo que a pasta de destino '{caminho_da_pasta_destino}' existe.")

        # 3. Salvar a lista original completa no arquivo files_list_original.csv
        caminho_completo_arquivo_original = os.path.join(caminho_da_pasta_destino, nome_arquivo_original)
        with open(caminho_completo_arquivo_original, 'w', encoding='utf-8') as f:
            f.write("filename,basename\n") # Adiciona o cabeçalho
            for filename, basename in files_list_original_data:
                f.write(f"{filename},{basename}\n")
        print(f"✅ Lista original completa ({len(files_list_original_data)} entradas) salva em '{caminho_completo_arquivo_original}' com sucesso!")

        # Embaralha a lista de arquivos para garantir uma divisão aleatória para teste/principal
        random.shuffle(todos_nomes_dos_arquivos)

        # 4. Calcular o número de arquivos para o conjunto de teste
        num_arquivos_teste = int(len(todos_nomes_dos_arquivos) * percentual_teste)

        # 5. Separar os arquivos em lista de teste e lista principal
        arquivos_teste = todos_nomes_dos_arquivos[:num_arquivos_teste]
        arquivos_principais = todos_nomes_dos_arquivos[num_arquivos_teste:]

        # 6. Salvar os nomes no arquivo de teste
        caminho_completo_arquivo_teste = os.path.join(caminho_da_pasta_destino, nome_arquivo_teste)
        with open(caminho_completo_arquivo_teste, 'w', encoding='utf-8') as f:
            f.write("filename,basename\n") # Adiciona o cabeçalho
            for nome in arquivos_teste:
                nome_sem_extensao = os.path.splitext(nome)[0]
                f.write(f"{nome},{nome_sem_extensao}\n")
        print(f"✅ Nomes dos arquivos de teste ({len(arquivos_teste)} arquivos) salvos em '{caminho_completo_arquivo_teste}' com sucesso!")

        # 7. Salvar os nomes no arquivo principal
        caminho_completo_arquivo_principal = os.path.join(caminho_da_pasta_destino, nome_arquivo_principal)
        with open(caminho_completo_arquivo_principal, 'w', encoding='utf-8') as f:
            f.write("filename,basename\n") # Adiciona o cabeçalho
            for nome in arquivos_principais:
                nome_sem_extensao = os.path.splitext(nome)[0]
                f.write(f"{nome},{nome_sem_extensao}\n")
        print(f"✅ Nomes dos arquivos principais ({len(arquivos_principais)} arquivos) salvos em '{caminho_completo_arquivo_principal}' com sucesso!")

    except FileNotFoundError:
        print(f"❌ Erro: A pasta de origem '{caminho_da_pasta_origem}' não foi encontrada. Verifique o caminho.")
    except Exception as e:
        print(f"❌ Ocorreu um erro: {e}")

# --- Uso da função ---
# Defina os caminhos para as pastas
caminho_origem = '/home/ematos/phd/pipeline_city/DATASET_CITY'
caminho_destino = '/home/ematos/phd/pipeline_city/DATASET_CITY_PROCESSED' # Esta pasta será criada se não existir

# Chame a função
criar_fila_de_nomes_txt(caminho_origem, caminho_destino)
```

```python
import os
import csv

def consolidar_linhas_de_arquivos(caminho_da_pasta_origem, caminho_da_pasta_processada,
                                  nome_arquivo_lista_original="files_list_original.csv",
                                  nome_arquivo_saida_consolidado="data_city_lines.csv",
                                  nome_arquivo_estatisticas="estatisticas_city.csv"): # Novo parâmetro para o nome do arquivo de estatísticas
    """
    Lê os nomes dos arquivos .txt de 'files_list_original.csv',
    abre cada um desses arquivos na pasta de origem,
    e consolida todas as suas linhas em um único arquivo CSV,
    adicionando um número de linha sequencial como a primeira coluna.
    Além disso, gera um arquivo de estatísticas com o total de linhas e palavras.

    Args:
        caminho_da_pasta_origem (str): O caminho para a pasta que contém os arquivos .txt originais (ex: DATASET_CITY).
        caminho_da_pasta_processada (str): O caminho para a pasta onde 'files_list_original.csv' está e onde
                                          'data_city_lines.csv' e 'estatisticas_city.csv' serão salvos (ex: DATASET_CITY_PROCESSED).
        nome_arquivo_lista_original (str): O nome do arquivo CSV que contém a lista original de filenames.
        nome_arquivo_saida_consolidado (str): O nome do arquivo CSV onde todas as linhas serão consolidadas.
        nome_arquivo_estatisticas (str): O nome do arquivo CSV onde as estatísticas serão salvas.
    """
    caminho_completo_lista_original = os.path.join(caminho_da_pasta_processada, nome_arquivo_lista_original)
    caminho_completo_arquivo_saida = os.path.join(caminho_da_pasta_processada, nome_arquivo_saida_consolidado)
    caminho_completo_arquivo_estatisticas = os.path.join(caminho_da_pasta_processada, nome_arquivo_estatisticas)
    
    linha_contador = 1 # Inicializa o contador de linhas para o arquivo de saída data_city_lines.csv
    total_linhas_consolidadas = 0 # Contador para o total de linhas em data_city_lines.csv
    total_palavras_consolidadas = 0 # Contador para o total de palavras em data_city_lines.csv

    try:
        # Garante que a pasta de destino exista
        os.makedirs(caminho_da_pasta_processada, exist_ok=True)
        print(f"📁 Garantindo que a pasta de destino '{caminho_da_pasta_processada}' existe.")

        # Abre o arquivo de saída consolidado no modo de escrita
        with open(caminho_completo_arquivo_saida, 'w', encoding='utf-8', newline='') as outfile:
            writer = csv.writer(outfile)
            writer.writerow(["line", "content"]) # Escreve o cabeçalho

            # Lê o arquivo files_list_original.csv
            with open(caminho_completo_lista_original, 'r', encoding='utf-8') as infile_list:
                reader = csv.reader(infile_list)
                next(reader) # Pula o cabeçalho do arquivo files_list_original.csv ("filename,basename")

                for row in reader:
                    if len(row) > 0:
                        filename_txt = row[0] # Pega o nome do arquivo .txt da primeira coluna
                        caminho_arquivo_txt = os.path.join(caminho_da_pasta_origem, filename_txt)

                        if os.path.exists(caminho_arquivo_txt):
                            print(f"Lendo arquivo: {filename_txt}")
                            with open(caminho_arquivo_txt, 'r', encoding='utf-8') as txt_file:
                                for line in txt_file:
                                    # Remove quebras de linha e espaços em branco extras
                                    cleaned_line = line.strip()
                                    if cleaned_line: # Garante que a linha não está vazia após a limpeza
                                        writer.writerow([linha_contador, cleaned_line])
                                        linha_contador += 1
                                        total_linhas_consolidadas += 1
                                        # Conta as palavras na linha (divide por espaço e remove vazios)
                                        total_palavras_consolidadas += len(cleaned_line.split())
                        else:
                            print(f"⚠️ Aviso: Arquivo '{caminho_arquivo_txt}' não encontrado. Pulando.")
            
        print(f"✅ Todas as linhas consolidadas em '{caminho_completo_arquivo_saida}' com sucesso!")

        # --------------------------------------------------------------------
        # Salva as estatísticas no arquivo estatisticas_city.csv
        with open(caminho_completo_arquivo_estatisticas, 'w', encoding='utf-8', newline='') as stats_file:
            stats_writer = csv.writer(stats_file)
            stats_writer.writerow(["Metrica", "Valor"]) # Cabeçalho do arquivo de estatísticas
            stats_writer.writerow(["Total de Palavras", total_palavras_consolidadas])
            stats_writer.writerow(["Total de Linhas", total_linhas_consolidadas])
        print(f"📊 Estatísticas salvas em '{caminho_completo_arquivo_estatisticas}' com sucesso!")
        # --------------------------------------------------------------------

    except FileNotFoundError:
        print(f"❌ Erro: O arquivo '{caminho_completo_lista_original}' ou a pasta de origem '{caminho_da_pasta_origem}' não foi encontrada. Verifique os caminhos.")
    except Exception as e:
        print(f"❌ Ocorreu um erro: {e}")

# --- Uso da função ---
# Defina os caminhos para as pastas
caminho_origem = '/home/ematos/phd/pipeline_city/DATASET_CITY'
caminho_destino_processado = '/home/ematos/phd/pipeline_city/DATASET_CITY_PROCESSED'

# Chame a função
consolidar_linhas_de_arquivos(caminho_origem, caminho_destino_processado)

```
